```python
## PART 1

d = open("input.txt").read().strip().split("\n\n")
ranges = d[0].split("\n")
ids = d[1].split("\n")
fresh = 0
for i in ids:
    for j in ranges:
        l, h = int(j.split("-")[0]), int(j.split("-")[1])
        if l <= int(i) <= h:
            # print(i)
            fresh += 1
            break
print(fresh)
```

```python
## PART 2

d = open("input.txt").read().strip().split("\n\n")
ranges = d[0].split("\n")

intervals = []
for r in ranges:
    l, h = r.split("-")
    intervals.append((int(l), int(h)))

intervals.sort(key=lambda x: x[0])

total = 0
if intervals:
    cur_l, cur_h = intervals[0]

    for l, h in intervals[1:]:
        if l <= cur_h:  
            if h > cur_h:
                cur_h = h
        else:
            total += cur_h - cur_l + 1
            cur_l, cur_h = l, h

    total += cur_h - cur_l + 1

print(total)
```
