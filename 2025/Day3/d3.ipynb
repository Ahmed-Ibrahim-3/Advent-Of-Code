```python
f = open("input.txt").read().strip().split("\n")

# f = [
#     "987654321111111",
#     "811111111111119",
#     "234234234234278",
#     "818181911112111"
# ]

sums = []

for line in f:
    digits = [c for c in line if c.isdigit()]
    best_val = -1
    best_pair = ""

    for i in range(len(digits)):
        for j in range(i + 1, len(digits)):
            val = int(digits[i]) * 10 + int(digits[j])
            if val > best_val:
                best_val = val
                best_pair = digits[i] + digits[j]

    print(best_pair)
    sums.append(int(best_pair))

print(sum(sums))
```

```python
f = open("input.txt").read().strip().split("\n")

# f = [ "987654321111111", "811111111111119", "234234234234278", "818181911112111" ]

def max_k_digits(line, k):
    digits = [c for c in line if c.isdigit()]
    if len(digits) <= k:
        return ''.join(digits)

    to_remove = len(digits) - k  
    stack = []

    for d in digits:
        while to_remove > 0 and stack and stack[-1] < d:
            stack.pop()
            to_remove -= 1
        stack.append(d)

    if to_remove > 0:
        stack = stack[:-to_remove]

    return ''.join(stack)

k = 12
sums = []

for line in f:
    best = max_k_digits(line, k)
    print(best)
    sums.append(int(best))

print(sum(sums))
```
