```python
## PART 1

f = open("input.txt", 'r')
rotations = f.read().strip().split('\n')

dial_position = 50
zeros = 0

for rotation in rotations:
    direction = rotation[0]

    distance = int(rotation[1:])

    if direction == 'R':
        dial_position = (dial_position + distance) % 100
    else:
        dial_position = (dial_position - distance) % 100

    if dial_position == 0:
        zeros += 1
```

```python
## PART 2

f = open("input.txt", 'r')
rotations = f.read().strip().split('\n')

dial_position = 50  
zeros = 0
N = 100              

for idx, rotation in enumerate(rotations):
    direction = rotation[0]
    distance = int(rotation[1:])

    step = 1 if direction == 'R' else -1

    for _ in range(distance):
        dial_position = (dial_position + step) % N
        if dial_position == 0:
            zeros += 1

    print("{3} | Rotation {0} : | Dial Pos : {1} | Zeros : {2}"
            .format(rotation, dial_position, zeros, idx))

print(zeros)
```
