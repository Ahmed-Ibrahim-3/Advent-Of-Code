```python
with open("input.txt", "r", encoding="utf-8") as f:
    grid_lines = [line.rstrip("\n") for line in f]

height = len(grid_lines)
width = len(grid_lines[0])

start_row = start_col = None
for r, row in enumerate(grid_lines):
    c = row.find('S')
    if c != -1:
        start_row, start_col = r, c
        break

splits = 0

active_beams = {start_col}

for r in range(start_row + 1, height):
    row = grid_lines[r]
    next_beams = set()

    for col in active_beams:
        if not (0 <= col < width):
            continue

        cell = row[col]

        if cell == '.':
            next_beams.add(col)

        elif cell == '^':
            splits += 1

            left = col - 1
            right = col + 1

            if left >= 0:
                next_beams.add(left)
            if right < width:
                next_beams.add(right)

        elif cell == 'S':
            next_beams.add(col)
        else:
            next_beams.add(col)

    active_beams = next_beams

print(splits)
```

```python

with open("input.txt", "r", encoding="utf-8") as f:
    grid_lines = [line.rstrip("\n") for line in f]

height = len(grid_lines)
width = len(grid_lines[0])

start_row = start_col = None
for r, row in enumerate(grid_lines):
    c = row.find('S')
    if c != -1:
        start_row, start_col = r, c
        break

dp = [0] * width
dp[start_col] = 1

for r in range(start_row, height - 1):
    row_below = grid_lines[r + 1]
    next_dp = [0] * width

    for c, ways in enumerate(dp):
        if ways == 0:
            continue

        cell = row_below[c]

        if cell == '^':
            if c - 1 >= 0:
                next_dp[c - 1] += ways
            if c + 1 < width:
                next_dp[c + 1] += ways
        else:
            next_dp[c] += ways

    dp = next_dp

print(sum(dp))
```
