```python
data = open("input.txt").read()
# data = """
# 123 328  51 64 
#  45 64  387 23 
#   6 98  215 314
# *   +   *   + """

lines = [line for line in data.strip().splitlines() if line.strip()]

*number_lines, op_line = lines

row_tokens = [line.split() for line in number_lines]
num_cols = len(row_tokens[0])

num_rows = len(number_lines)
row_len = max(len(line) for line in number_lines)

num_grid = [line.ljust(row_len) for line in number_lines]
op_line = op_line.ljust(row_len)

op_positions = [i for i, ch in enumerate(op_line) if ch in "+*"]

total = 0

for idx, j in enumerate(op_positions):

    jnext = op_positions[idx + 1] if idx + 1 < len(op_positions) else row_len

    col_numbers = []

    for c in range(jnext - 1, j - 1, -1):
        digits = []
        for r in range(num_rows):
            ch = num_grid[r][c]
            if ch.isdigit():
                digits.append(ch)
        if digits:
            col_numbers.append(int("".join(digits)))

    op = op_line[j]
    if op == "+":
        total += sum(col_numbers)
    elif op == "*":
        prod = 1
        for v in col_numbers:
            prod *= v
        total += prod

print(total)
```
