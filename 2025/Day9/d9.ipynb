```python
from collections import deque
from itertools import combinations


tiles = []
with open("input.txt") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        x, y = map(int, line.split(","))
        tiles.append((x, y))

points = tiles[:]  
n = len(points)


xs = [x for x, _ in points]
ys = [y for _, y in points]

min_x, max_x = min(xs), max(xs)
min_y, max_y = min(ys), max(ys)

X_lines = sorted(set(xs + [min_x - 1, max_x + 1]))
Y_lines = sorted(set(ys + [min_y - 1, max_y + 1]))

Nx, Ny = len(X_lines), len(Y_lines)
x_id = {x: i for i, x in enumerate(X_lines)}
y_id = {y: i for i, y in enumerate(Y_lines)}

vert_wall = [[False] * (Ny - 1) for _ in range(Nx)]
horiz_wall = [[False] * Ny for _ in range(Nx - 1)]

for i in range(n):
    x1, y1 = points[i]
    x2, y2 = points[(i + 1) % n]

    if x1 == x2:
        xi = x_id[x1]
        y1i, y2i = y_id[y1], y_id[y2]
        if y1i > y2i:
            y1i, y2i = y2i, y1i
        for j in range(y1i, y2i):
            vert_wall[xi][j] = True
    elif y1 == y2:
        yi = y_id[y1]
        x1i, x2i = x_id[x1], x_id[x2]
        if x1i > x2i:
            x1i, x2i = x2i, x1i
        for ii in range(x1i, x2i):
            horiz_wall[ii][yi] = True
    
W = Nx - 1
H = Ny - 1

outside = [[False] * H for _ in range(W)]
dq = deque()

outside[0][0] = True
dq.append((0, 0))

while dq:
    i, j = dq.popleft()

    if i > 0 and not outside[i - 1][j] and not vert_wall[i][j]:
        outside[i - 1][j] = True
        dq.append((i - 1, j))

    if i + 1 < W and not outside[i + 1][j] and not vert_wall[i + 1][j]:
        outside[i + 1][j] = True
        dq.append((i + 1, j))

    if j > 0 and not outside[i][j - 1] and not horiz_wall[i][j]:
        outside[i][j - 1] = True
        dq.append((i, j - 1))

    if j + 1 < H and not outside[i][j + 1] and not horiz_wall[i][j + 1]:
        outside[i][j + 1] = True
        dq.append((i, j + 1))

inside = [[0] * H for _ in range(W)]
for i in range(W):
    for j in range(H):
        inside[i][j] = 0 if outside[i][j] else 1

pref = [[0] * (H + 1) for _ in range(W + 1)]
for i in range(W):
    row_sum = 0
    for j in range(H):
        row_sum += inside[i][j]
        pref[i + 1][j + 1] = pref[i][j + 1] + row_sum

def cells_sum(ix1, iy1, ix2, iy2):
    return (
        pref[ix2 + 1][iy2 + 1]
        - pref[ix1][iy2 + 1]
        - pref[ix2 + 1][iy1]
        + pref[ix1][iy1]
    )

max_area = 0
best_pair = None

for (x1, y1), (x2, y2) in combinations(points, 2):
    if x1 == x2 or y1 == y2:
        continue

    x_lo, x_hi = sorted((x1, x2))
    y_lo, y_hi = sorted((y1, y2))

    ix_lo = x_id[x_lo]
    ix_hi = x_id[x_hi]
    iy_lo = y_id[y_lo]
    iy_hi = y_id[y_hi]

    cells_w = ix_hi - ix_lo
    cells_h = iy_hi - iy_lo
    if cells_w <= 0 or cells_h <= 0:
        continue

    inside_cells = cells_sum(ix_lo, iy_lo, ix_hi - 1, iy_hi - 1)
    if inside_cells != cells_w * cells_h:
        continue

    area = (x_hi - x_lo + 1) * (y_hi - y_lo + 1)
    if area > max_area:
        max_area = area
        best_pair = ((x1, y1), (x2, y2))

print(max_area)

```
