```python
# graph = """
# aaa: you hhh
# you: bbb ccc
# bbb: ddd eee
# ccc: ddd eee fff
# ddd: ggg
# eee: out
# fff: out
# ggg: out
# hhh: ccc fff iii
# iii: out
# """
graph = open("input.txt").read()


def parse_graph(text):
    graph = {}
    for line in text.strip().splitlines():
        left, right = line.split(':', 1)
        node = left.strip()
        targets = right.split()
        graph[node] = targets
    return graph

def count_paths(graph, start, end="out"):
    def dfs(node, visited):
        if node == end:
            return 1  
        if node in visited:
            return 0  
        visited.add(node)
        total = 0
        for nxt in graph.get(node, []):
            total += dfs(nxt, visited)
        visited.remove(node)
        return total

    return dfs(start, set())

def list_paths(graph, start, end="out"):
    all_paths = []

    def dfs(node, path, visited):
        if node in visited:
            return
        path.append(node)
        visited.add(node)

        if node == end:
            all_paths.append(path.copy())
        else:
            for nxt in graph.get(node, []):
                dfs(nxt, path, visited)

        path.pop()
        visited.remove(node)

    dfs(start, [], set())
    return all_paths


graph = parse_graph(graph)

start_node = "you"
total = count_paths(graph, start_node, "out")
paths = list_paths(graph, start_node, "out")

print("Total paths from", start_node, "to out:", total)
for p in paths:
    print(" -> ".join(p))
```

```python
from functools import lru_cache

def parse_graph(text):
    graph = {}
    for line in text.strip().splitlines():
        line = line.strip()
        if not line:
            continue
        left, right = line.split(":", 1)
        node = left.strip()
        targets = right.strip().split()
        graph[node] = targets
    return graph

# graph = """
# svr: aaa bbb
# aaa: fft
# fft: ccc
# bbb: tty
# tty: ccc
# ccc: ddd eee
# ddd: hub
# hub: fff
# eee: dac
# dac: fff
# fff: ggg hhh
# ggg: out
# hhh: out
# """
graph = open("input.txt").read()


graph = parse_graph(graph)

START = "svr"
END = "out"
REQ1 = "dac"
REQ2 = "fft"

@lru_cache(maxsize=None)
def count_all_paths(node):
    if node == END:
        return 1
    total = 0
    for nxt in graph.get(node, []):
        total += count_all_paths(nxt)
    return total

@lru_cache(maxsize=None)
def count_paths_with_both(node, visited_dac, visited_fft):
    visited_dac = visited_dac or (node == REQ1)
    visited_fft = visited_fft or (node == REQ2)

    if node == END:
        return int(visited_dac and visited_fft)

    total = 0
    for nxt in graph.get(node, []):
        total += count_paths_with_both(nxt, visited_dac, visited_fft)
    return total

total_paths = count_all_paths(START)
paths_with_dac_and_fft = count_paths_with_both(START, False, False)

print("Total paths from svr to out:", total_paths)
print("Paths from svr to out that visit both dac and fft:", paths_with_dac_and_fft)
```
